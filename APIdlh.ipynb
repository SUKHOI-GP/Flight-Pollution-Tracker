```python
import requests
```

```python
url = "https://opensky-network.org/api/states/all"
params = {
    "lamin": 47, 
    "lamax": 55,
    "lomin": 5, 
    "lomax": 15
}
```

```python
response = requests.get(url, params=params)
```

```python
if response.status_code == 200:
    data = response.json()
    fluge_dlh = [s for s in data["states"] if s[1] and s[1].startswith("DLH")]
    print("Lufhansa über Deutschland")
    for flug in fluge_dlh:
        callsign = flug[1]
        latitud = flug[6]
        longitud = flug[5]
        altitud = flug[7]
        print(f"{callsign}: lat={latitud}, lon={longitud}, alt={altitud}")
else:
        print("Error:", response.status_code)
```
